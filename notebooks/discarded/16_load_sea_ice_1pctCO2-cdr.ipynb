```python
si_data = {model:{} for model in models}
for model in models:
    model_dir = os.path.join(ramp_down_dir, model,'1pctCO2-cdr')
    print(model)
    print(os.listdir(model_dir))
    si_data[model]['1pctCO2-cdr'] = xr.open_mfdataset(
        os.path.join(model_dir,'*.nc'),
        combine="nested",
        concat_dim="time",
        use_cftime=True,
        parallel=True,
        chunks={'time':100},
        ).siconc.isel(time=slice(None,12*340))
```

```python
path = "/g/data/if69/sc1326/CDRMIP/data/raw/siconc/1pctCO2-cdr/*MIROC-ES2L_*.nc"
si_data['MIROC-ES2L']['1pctCO2-cdr'] = xr.open_mfdataset(
    path,
    combine='by_coords',
    use_cftime=True,
    parallel=True,
    chunks={'time':100},
).load()
```

```python
# save
save_dir = os.path.join(data_dir,'raw/siconc/1pctCO2-cdr')
for model, ds_dict in si_data.items():
    if model not in ['MIROC-ES2L']:
        continue
    ds = ds_dict['1pctCO2-cdr'].load()
    path = os.path.join(
        save_dir,
        f"{model}_1pctCO2-cdr_siconc.nc"
    )
    ds.to_netcdf(path)
```
